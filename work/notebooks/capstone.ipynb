## Week-8 demo outline (5 minutes)

**0:00 to 0:40. The question.**
FlyRank editors cannot review every page. Which pages should they read first when search impressions start to fall? Frame it as decision support, not prediction.

**0:40 to 1:40. The data and the label.**
Anonymized FlyRank slice: 30,000 pages, 32 clients. I kept pages with at least 50 impressions in the previous 30 days, which left 20,249 pages from 30 clients. Label: "declined" means the last 30 days fell below 70% of the previous 30 days. The base rate is 0.521, so about half of pages declined, and that is the number every result must beat.

**1:40 to 2:40. The method.**
Rule baseline from Week 4 vs a logistic regression (balanced classes). Validation holds out whole clients (GroupKFold on `client_id`), so a client never sits on both sides of the split. I left out the trend fields because the label comes from them.

**2:40 to 3:40. One chart: decline rate by content age.**
Show `work/figures/decline_by_age.png`. Younger pages declined more often (0.594, 0.691, 0.562) than older ones (0.446, 0.308). Takeaway in one sentence: in this sample, decline tracks page age.

**3:40 to 4:20. One honest result.**
Mean AUC over 10 client-held-out splits: model 0.583, rule 0.420. Directionally better, but weak and unstable (range 0.455 to 0.876), and without content age the AUC drops to 0.475. The label is an observed drop over one window, so nothing here shows that a refresh would help.

**4:20 to 5:00. One recommendation.**
Use the queue as a reading list for a human editor, starting with the top 100 to 200 picks, where the decline rate was about 0.66 against a 0.538 base. Do not automate changes from the score. Re-validate on held-out clients before using it on a new client.

## Shareable cut 1: social post (methodology)

I built a ranked "read these pages first" queue for content editors, then tried to break my own result.

The trap: if pages from the same client land in both training and test, the score looks better than it is. So I held out whole clients. Result on anonymized FlyRank search data (20,249 pages, 30 clients): AUC 0.583 vs 0.420 for a simple rule. Better than the rule, but weak, and most of the signal came from one feature, page age.

So I framed it as decision support for a human editor, not an automatic fix. A small signal, honestly measured, is still worth reporting.
